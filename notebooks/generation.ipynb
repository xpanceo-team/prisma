```python
import torch

from crystal_diffusers.pipelines.mattergen import MatterGenPipeline
```

```python
pipeline = MatterGenPipeline.from_pretrained("atom-int-team/mattergen-formula-e_above_hull", use_safetensors=False)
```

```python
device = "cuda"

batch_size = 2
random_seed = 42

condition = {
    "e_above_hull": torch.tensor([0.0430, 0.0430]),
    'formula': ['Na3 Mn1 Co1 Ni1 O6', 'Na3 Mn1 Co1 Ni1 O6']
}

# deterministic generation
generator = torch.Generator(device=device).manual_seed(random_seed)
```

```python
structures = pipeline(
    batch_size=batch_size, 
    condition=condition,
    guidance_scale=3.0,
    device=device, 
    generator=generator
)
```
